#ESTRUCTURA DE REPETICIÓN
Estructura computacional que permite repetir una acción n cantidad de veces o hasta que se logre una condición.


##Estructura PARA
(FOR)
Estructura empleada cuando se sabe cuántas veces se debe repetir una acción.

Sintaxis general
```
for (variable) in range( VI,VF+1,Paso): Aquí se repite (VF-VI)/Paso veces
```
Donde:
* VI: Valor inicial
* VF: Valor final
* Paso

Sintaxis resumida
```
# Cuando no me interesa el valor inicial
for (variable) in range(VF):
  Se repite VF veces.
```



````python
#Hacer un programa que imprima los números pares del 2 al 20
for par in range(2,21,2):
  print(par)
````

````python
for par in range(2,21):
  if par %2==0:
    print(par)
````

````python
#Tabla de multiplicar del 5
print("Tabla del 5")
for i in range (1, 11, 1):
  print(f"5 x {i} = {5*i}")
````

````python
#for en cascada
for x in range (1,11): #Este controla el número de la tabla
  print(f"Tabla del {x}")
  for y in range (1,11): #Este controla la tabla
    print(f"{x} x {y}={x*y}")
  print()
````

````python
n=int(input("Hasta qué numero de serie desea visualizar?"))
for serie in range(1,n+1):
  if serie==1: print(1, end=" ")
  elif serie%2==0: print(f"- 1/{serie}", end=" ")
  else: print(f"+ 1/{serie}", end=" ")
````

````python
#Hacer un programa que imprima los números pares del 2 al 20
par=2
while par <=20:
  print(par)
  par=par+2 #par+=2
````

````python
#Tabla de multiplicar del 5
i=1
print("Tabla del 5")
while i<=10:
  print(f"5 x {i} = {5*i}")
  i=i+1
````

````python
#Problemas con el espacio de indentación, tenerlo en cuenta para el resto de ejercicios
x=1
while x<=10:
  print(f"Tabla del {x}")
  y=1
  while y<=10:
    print(f"{x} x {y} = {x*y}")
    y=y+1
  x=x+1
  print()
````

````python
t=1
while t<=10:
  print(f"Tabla del {t}")
  y=1
  while y<=10:
    print(f"{t} x {y} = {t*y}")
    y=y+1
  t=t+1
  print()
````
